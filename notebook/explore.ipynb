```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

from trader.learn import generate_trending_sinusoid, subsample_data
```

```python
print("Generating and plotting synthetic sinusoidal data...")
synthetic_df = generate_trending_sinusoid(num_days=355) # Using 30 days for a clearer plot
synthetic_df = subsample_data(synthetic_df, freq='D') # Subsample for clarity

display(synthetic_df)
```

```python
plt.plot(synthetic_df['timestamp'], synthetic_df['close_eur'])
plt.title('Synthetic Trending Sinusoidal Price Data')
plt.xlabel('Timestamp')
plt.ylabel('Price')
plt.grid(True)
plt.show()
```
